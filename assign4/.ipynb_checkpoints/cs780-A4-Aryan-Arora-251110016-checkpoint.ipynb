# Helper Functions
<a id="helper"></a>

Write all the helper functions that will be used for value-based and policy based algorithms below. In case you want to add more helper functions, please feel free to add.

````python
def selectGreedyAction(net, state):
    with torch.no_grad():
        Q = net(torch.FloatTensor(state)).squeeze(0)
    ls = [a for a in range(Q.shape[0]) if Q[a] == torch.max(Q)]
    greedyAction = np.random.choice(ls)
    return greedyAction
````

````python
def selectEpsilonGreedyAction(net,  state, epsilon):
    #this function gets q-values via the network and selects an action from q-values using epsilon greedy strategy
    #and returns it
    #note this function can be used for decaying epsilon greedy strategy,
    #you would need to create a wrapper function that will handle decaying epsilon
    #you can create this wrapper in this helper function section
    #for the agents you would be implementing it would be nice to play with decaying parameter to get optimal results

    #Your code goes in here
    with torch.no_grad():
        Q = net(torch.FloatTensor(state)).squeeze(0)
    if np.random.rand() < epsilon:
      eGreedyAction = np.random.choice(range(Q.shape[0]))
    else:
      ls = [a for a in range(Q.shape[0]) if Q[a] == torch.max(Q)]
      eGreedyAction = np.random.choice(ls)

    return eGreedyAction
````

````python
def selectSoftMaxAction(net, state, temp):
    #this function gets q-values via the network and selects an action from q-values using softmax strategy
    #and returns it
    #note this function can be used for decaying temperature softmax strategy,
    #you would need to create a wrapper function that will handle decaying temperature
    #you can create this wrapper in this helper function section
    #for the agents you would be implementing it would be nice to play with decaying parameter to get optimal results

    #Your code goes in here
    with torch.no_grad():
        Q = net(torch.FloatTensor(state)).squeeze(0)
    p = F.softmax(Q/temp, dim=0)
    softAction = np.random.choice(range(Q.shape[0]), p=p)

    return softAction
````

````python
#Value Network
def createValueNetwork(inDim, outDim, hDim = [32,32], activation = F.relu):
    #this creates a Feed Forward Neural Network class and instantiates it and returns the class
    #the class should be derived from torch nn.Module and it should have init and forward method at the very least
    #the forward function should return q-value for each possible action
    class ValueNetwork(nn.Module):
      def __init__(self, inDim, outDim, hDim = [32,32], activation = F.relu):
        super().__init__()
        n = len(hDim)
        hDim = [inDim] + hDim + [outDim]
        layers = []
        for i in range(n + 1):
          layers.append(nn.Linear(hDim[i], hDim[i + 1]))
        self.layers = nn.ModuleList(layers)
        self.activation = activation

      def forward(self, state):
        for i in range(len(self.layers)-1):
          state = self.activation(self.layers[i](state))

        state = self.layers[-1](state)
        return state

    valueNetwork = ValueNetwork(inDim, outDim, hDim, activation)
    return valueNetwork
````

````python
#Dueling Network
def createDuelingNetwork(inDim, outDim, hDim = [32,32], activation = F.relu):
    class DuelNetwork(nn.Module):
      def __init__(self, inDim, outDim, hDim, activation):
        super().__init__()
        n = len(hDim)
        hDim = [inDim] + hDim
        layers = []
        for i in range(n):
          layers.append(nn.Linear(hDim[i], hDim[i+1]))
        self.v_layer = nn.Linear(hDim[-1], 1)
        self.a_layer = nn.Linear(hDim[-1], outDim)
        self.layers = nn.ModuleList(layers)
        self.activation = activation
      def forward(self,state):
        if state.dim() == 1:
          state = state.unsqueeze(0)
        for layer in self.layers:
          state = self.activation(layer(state))
        v = self.v_layer(state)
        a = self.a_layer(state)
        return v + (a - torch.mean(a,axis=1,keepdim=True))
    duelNetwork = DuelNetwork(inDim, outDim, hDim, activation)
    return duelNetwork
````

````python
#Policy Network
def createPolicyNetwork(inDim, outDim, hDim = [32,32], activation = F.relu):
    class PolicyNetwork(nn.Module):
      def __init__(self, inDim, outDim, hDim = [32,32], activation = F.relu):
        super().__init__()
        n = len(hDim)
        hDim = [inDim] + hDim + [outDim]
        layers = []
        for i in range(n + 1):
          layers.append(nn.Linear(hDim[i], hDim[i + 1]))
        self.layers = nn.ModuleList(layers)
        self.activation = activation

      def forward(self, state):
        for i in range(len(self.layers)-1):
          state = self.activation(self.layers[i](state))

        state = self.layers[-1](state)
        return state

    policyNetwork = PolicyNetwork(inDim, outDim, hDim, activation)

    return policyNetwork
````

````python
def huberLoss(error, delta):
    #this function calculates the huber loss for the error using the delta parameter

    #Your code goes in here
    loss = torch.where(torch.abs(error) <= delta, 0.5 * error ** 2, delta * (abs(error) - 0.5 * delta))
    return loss
````

## ReplayBuffer (50 Marks)

In next few cells, you will implement replaybuffer class.

This class creates a buffer for storing and retrieving experiences. This is a generic class and can be used
for different agents like NFQ, DQN, DDQN, PER_DDQN, etc.
Following are the methods for this class which are implemented in subsequent cells

```
class ReplayBuffer():
    def __init__(self, bufferSize, **kwargs)
    def store(self, experience)
    def update(self, indices, priorities)
    def collectExperiences(env, state, explorationStrategy, net = None)
    def sample(self, batchSize, **kwargs)
    def splitExperiences(self, experiences)
    def length(self)
```   

````python
class ReplayBuffer():
    def __init__(self, bufferSize, bufferType = 'DQN', **kwargs):
        # this function creates the relevant data-structures, and intializes all relevant variables
        # it can take variable number of parameters like alpha, beta, beta_rate (required for PER)
        # here the bufferType variable can be used to maintain one class for all types of agents
        # using the bufferType parameter in the methods below, you can implement all possible functionalities
        # that could be used for different types of agents
        # permissible values for bufferType = NFQ, DQN, DDQN, D3QN and PER-D3QN

        #Your code goes in here
        self.buffer = deque(maxlen=bufferSize)
        self.bufferSize= bufferSize
        self.bufferType = bufferType
        if bufferType == 'PER-D3QN':
            self.alpha = kwargs['alpha']
            self.beta  = kwargs['beta']
            self.beta_rate = kwargs['beta_rate']
            self.priorities = deque(maxlen=bufferSize)
        return
````

````python
class ReplayBuffer(ReplayBuffer):
    def store(self, experience):
        self.buffer.append(experience)
        if self.bufferType == 'PER-D3QN':
            maxp = max(self.priorities) if self.priorities else 1.0
            self.priorities.append(maxp)
        return
````

````python
class ReplayBuffer(ReplayBuffer):
    def update(self, indices, priorities):
        for i,idx in enumerate(indices):
          self.priorities[idx] = priorities[i]
        return
````

````python
class ReplayBuffer(ReplayBuffer):
    def collectExperiences(self,env, state, explorationStrategy, countExperiences, net = None):
        for _ in range(countExperiences):
            action = explorationStrategy(net, state)
            s, r, terminated, truncated, info = env.step(action)
            done = terminated or truncated
            experience = (state,action,r,s,done)
            self.store(experience)
            state = s
            if done:
                break
        return
````

````python
class ReplayBuffer(ReplayBuffer):
    def sample(self, batchSize, **kwargs):
        if self.bufferType == 'PER-D3QN':
            priorities = np.array(self.priorities)
            p = priorities**self.alpha
            p = p/p.sum()
            inds = np.random.choice(len(self.buffer), size=batchSize, p=p)
            experiencesList=[self.buffer[i] for i in inds]
            w = (len(self.buffer)*p[inds])**(-self.beta)
            w = w/w.max()
            self.beta = min(1.0, self.beta_rate + self.beta)
            return experiencesList, inds, torch.FloatTensor(w)
            
        k = min(batchSize, len(self.buffer))
        experiencesList = random.sample(self.buffer,k=k)


        return experiencesList
````

````python
class ReplayBuffer(ReplayBuffer):
    def splitExperiences(self, experiences):
        #it takes in experiences and gives the following:
        #states, actions, rewards, nextStates, dones
        #
        #Your code goes in here
        #
        states = torch.FloatTensor(np.array([experiences[i][0] for i in range(len(experiences))]))
        actions = torch.LongTensor(np.array([experiences[i][1] for i in range(len(experiences))]))
        rewards = torch.FloatTensor(np.array([experiences[i][2] for i in range(len(experiences))]))
        nextStates = torch.FloatTensor(np.array([experiences[i][3] for i in range(len(experiences))]))
        dones = torch.FloatTensor(np.array([experiences[i][4] for i in range(len(experiences))]))
        return states, actions, rewards, nextStates, dones
````

````python
class ReplayBuffer(ReplayBuffer):
    def length(self):
        bufferSize = len(self.buffer)
        return bufferSize
````

````python
class NFQ():
    def __init__(self, env, seed, gamma, epochs,
                 bufferSize,
                 batchSize,
                 optimizerFn,
                 optimizerLR,
                 MAX_TRAIN_EPISODES, MAX_EVAL_EPISODES,
                 explorationStrategyTrainFn,
                 explorationStrategyEvalFn, loss=F.mse_loss):
    
        self.env = env
        self.seed = seed
        self.loss = loss
        self.env.reset(seed=seed)
        self.net = createValueNetwork(env.observation_space.shape[0], env.action_space.n)
        self.gamma = gamma
        self.epochs = epochs
        self.buffer = ReplayBuffer(bufferSize)
        self.strategyTrain = explorationStrategyTrainFn
        self.strategyEval = explorationStrategyEvalFn
        self.maxTrainEpisodes = MAX_TRAIN_EPISODES
        self.maxEvalEpisodes = MAX_EVAL_EPISODES
        self.optimizer = optimizerFn(self.net.parameters(), lr=optimizerLR)
        self.batchSize = batchSize
        self.initBookKeeping()

````

````python
class NFQ(NFQ):
    def initBookKeeping(self):
        #this method creates and intializes all the variables required for book-keeping values and it is called
        #init method
        self.trainRewards = []
        self.evalRewards = []
        self.trainTime = []
        self.wallTime = []
        self.totalSteps = []
        self.stepCount = 0
        self.currentEpisodeReward = 0
        self.currentEpisodeSteps = 0
        self.currentTrainTime = 0
        self.currentWallTime = 0
        self.currentEvalReward = 0
````

````python
class NFQ(NFQ):
    def runNFQ(self):
        #this is the main method, it trains the agent, performs bookkeeping while training and finally evaluates
        #the agent and returns the following quantities:
        #1. episode wise mean train rewards
        #2. epsiode wise mean eval rewards
        #2. episode wise trainTime (in seconds): time elapsed during training since the start of the first episode
        #3. episode wise wallClockTime (in seconds): actual time elapsed since the start of training,
        #                               note this will include time for BookKeeping and evaluation
        # Note both trainTime and wallClockTime get accumulated as episodes proceed.


        #Your code goes in here
        trainRewardsList, trainTimeList, evalRewardsList, wallClockTimeList = self.trainAgent()
        finalEvalReward = evalRewardsList[-1]
        
        return trainRewardsList, trainTimeList, evalRewardsList, wallClockTimeList, finalEvalReward
````

````python
class NFQ(NFQ):
    def trainAgent(self):
        #this method collects experiences and trains the NFQ agent and does BookKeeping while training.
        #this calls the trainNetwork() method internally, it also evaluates the agent per episode
        #it trains the agent for MAX_TRAIN_EPISODES

        #Your code goes in here
        evalRewardsList = []
        self.currentWallTime = 0
        self.currentTrainTime = 0
        for e in range(self.maxTrainEpisodes):
            wallStart = time.time()
            self.buffer.buffer.clear()
            state, info = self.env.reset(seed=self.seed+e)
            self.buffer.collectExperiences(self.env,state,self.strategyTrain, self.batchSize, self.net)
            experiencesList = self.buffer.sample(self.batchSize)
            _, _, rewards, _, _  = self.buffer.splitExperiences(experiencesList)
            self.stepCount+= len(experiencesList)
            self.currentEpisodeReward = rewards.mean().item()
            self.currentEpisodeSteps = len(experiencesList)
            trainStart = time.time()
            self.trainNetwork(experiencesList,self.epochs)
            trainEnd = time.time()
            self.currentTrainTime += trainEnd - trainStart
            ls = np.array(self.evaluateAgent()[-self.maxEvalEpisodes:])
            evalRewardsList.append(ls.mean())
            wallEnd = time.time()
            self.currentWallTime += wallEnd-wallStart
            self.performBookKeeping(train=True)
            
        trainRewardsList=self.trainRewards
        trainTimeList = self.trainTime
        wallClockTimeList = self.wallTime
        
        return trainRewardsList, trainTimeList, evalRewardsList, wallClockTimeList
````

````python
class NFQ(NFQ):
    def trainNetwork(self, experiences, epochs):
        states,actions,rewards,nextStates,dones =  self.buffer.splitExperiences(experiences)
        for _ in range(epochs):
            with torch.no_grad():
                Q_Vals_next = self.net(nextStates)
                targets = torch.FloatTensor(rewards+ ((self.gamma*(torch.max(Q_Vals_next,dim=1).values)*(1-dones))))
            Q_Vals = self.net(states)
            Q_Vals = Q_Vals.gather(1, actions.unsqueeze(1)).squeeze(1)
            loss = self.loss(Q_Vals,targets)
            self.optimizer.zero_grad()
            loss.backward()
            self.optimizer.step()
        
#ch number of times and is called by the trainAgent function
        # it essentially uses the experiences to calculate target, using the targets it calculates the error, which
        # is further used for calulating the loss. It then uses the optimizer over the loss
        # to update the params of the network by backpropagating through the network
        # this function does not return anything
        # you can try out other loss functions other than MSE like Huber loss, MAE, etc.
````

````python
class NFQ(NFQ):
    def evaluateAgent(self):
        #this function evaluates the agent using the value network, it evaluates agent for MAX_EVAL_EPISODES
        #typcially MAX_EVAL_EPISODES = 1
        for i in range(self.maxEvalEpisodes):
            state, info = self.env.reset(seed=self.seed + i)
            self.currentEvalReward = 0
            done = False
            while not done:
                action = self.strategyEval(self.net, state)
                state, r, term, trunc, _ = self.env.step(action)
                done = term or trunc
                self.currentEvalReward += r
            self.performBookKeeping(train=False)
            
        finalEvalRewardsList = self.evalRewards
        return finalEvalRewardsList
````

````python
class DDQN(DDQN):
    def trainAgent(self):
        self.updateNetwork(self.onlineNet, self.targetNet)
        evalRewardsList = []
        self.currentTrainTime = 0
        self.currentWallTime = 0
        for e in range(self.maxTrainEpisodes):
            wallStart = time.time()
            state, info = self.env.reset(seed=self.seed+e)
            done = False
            self.currentEpisodeReward = 0
            self.currentEpisodeSteps = 0
            while not done: 
                action = self.strategyTrain(self.onlineNet,state)
                nextState, reward, term, trunc, _ = self.env.step(action)
                done = term or trunc
                self.buffer.store((state,action,reward,nextState,done))
                state = nextState
                self.currentEpisodeReward += reward
                self.currentEpisodeSteps +=1 
                self.stepCount+=1
                if self.buffer.length() >= self.batchSize:
                    experiences = self.buffer.sample(self.batchSize)
                    trainStart = time.time()
                    self.trainNetwork(experiences)
                    trainEnd = time.time()
                    self.currentTrainTime += trainEnd - trainStart         
                if self.stepCount % self.updateFrequency == 0:
                    self.updateNetwork(self.onlineNet, self.targetNet)
            self.currentEpisodeReward /=self.currentEpisodeSteps
            ls = np.array(self.evaluateAgent()[-self.maxEvalEpisodes:])
            evalRewardsList.append(ls.mean())
            wallEnd = time.time()
            self.currentWallTime += wallEnd-wallStart
            self.performBookKeeping(train=True)
            
        trainRewardsList=self.trainRewards
        trainTimeList = self.trainTime
        wallClockTimeList = self.wallTime
        
        return trainRewardsList, trainTimeList, evalRewardsList, wallClockTimeList
````

````python
class DDQN(DDQN):
    def trainNetwork(self, experiences, epochs=1):
        states, actions, rewards, nextStates, dones = self.buffer.splitExperiences(experiences)

        with torch.no_grad():
            x = self.onlineNet(nextStates).argmax(dim=1)
            Q_next = self.targetNet(nextStates)
            mQ_next = Q_next.gather(1, x.unsqueeze(1)).squeeze(1)
            targets = rewards + self.gamma * mQ_next * (1 - dones)

        Q_vals = self.onlineNet(states)
        Q_vals = Q_vals.gather(1, actions.unsqueeze(1)).squeeze(1)

        loss = self.loss(Q_vals, targets)
        self.optimizer.zero_grad()
        loss.backward()
        self.optimizer.step()
        return
````

````python
class DDQN(DDQN):
    def updateNetwork(self, onlineNet, targetNet):
        targetNet.load_state_dict(onlineNet.state_dict())

        return
````

````python
class DDQN(DDQN):
    def evaluateAgent(self):
        for i in range(self.maxEvalEpisodes):
            state, info = self.env.reset(seed=self.seed + i)
            self.currentEvalReward = 0
            done = False
            while not done:
                action = self.strategyEval(self.onlineNet, state)
                state, r, term, trunc, _ = self.env.step(action)
                done = term or trunc
                self.currentEvalReward += r
            self.performBookKeeping(train=False)
            
        finalEvalRewardsList = self.evalRewards
        return finalEvalRewardsList
````

````python
class D3QN(D3QN):
    def performBookKeeping(self, train = True):
        if train:
          self.trainRewards.append(self.currentEpisodeReward)
          self.trainTime.append(self.currentTrainTime)
          self.wallTime.append(self.currentWallTime)
          self.totalSteps.append(self.currentEpisodeSteps)
        else:
          self.evalRewards.append(self.currentEvalReward)
        return
````

````python
class D3QN(D3QN):
    def runD3QN(self):
        #this is the main method, it trains the agent, performs bookkeeping while training and finally evaluates
        #the agent and returns the following quantities:
        #1. episode wise mean train rewards
        #2. epsiode wise mean eval rewards
        #2. episode wise trainTime (in seconds): time elapsed during training since the start of the first episode
        #3. episode wise wallClockTime (in seconds): actual time elapsed since the start of training,
        #                               note this will include time for BookKeeping and evaluation
        # Note both trainTime and wallClockTime get accumulated as episodes proceed.

        #Your code goes in here
        trainRewardsList, trainTimeList, evalRewardsList, wallClockTimeList = self.trainAgent()
        finalEvalReward = evalRewardsList[-1]
        return trainRewardsList, trainTimeList, evalRewardsList, wallClockTimeList, finalEvalReward
````

````python
class D3QN(D3QN):
    def trainAgent(self):
        self.updateNetwork(self.onlineNet, self.targetNet)
        evalRewardsList = []
        self.currentTrainTime = 0
        self.currentWallTime = 0
        for e in range(self.maxTrainEpisodes):
            wallStart = time.time()
            state, info = self.env.reset(seed=self.seed+e)
            done = False
            self.currentEpisodeReward = 0
            self.currentEpisodeSteps = 0
            while not done: 
                action = self.strategyTrain(self.onlineNet,state)
                nextState, reward, term, trunc, _ = self.env.step(action)
                done = term or trunc
                self.buffer.store((state,action,reward,nextState,done))
                state = nextState
                self.currentEpisodeReward += reward
                self.currentEpisodeSteps +=1 
                self.stepCount+=1
                if self.buffer.length() >= self.batchSize:
                    experiences = self.buffer.sample(self.batchSize)
                    trainStart = time.time()
                    self.trainNetwork(experiences)
                    trainEnd = time.time()
                    self.currentTrainTime += trainEnd - trainStart         
                if self.stepCount % self.updateFrequency == 0:
                    self.updateNetwork(self.onlineNet, self.targetNet)
            self.currentEpisodeReward /=self.currentEpisodeSteps
            ls = np.array(self.evaluateAgent()[-self.maxEvalEpisodes:])
            evalRewardsList.append(ls.mean())
            wallEnd = time.time()
            self.currentWallTime += wallEnd-wallStart
            self.performBookKeeping(train=True)
            
        trainRewardsList=self.trainRewards
        trainTimeList = self.trainTime
        wallClockTimeList = self.wallTime
        
        return trainRewardsList, trainTimeList, evalRewardsList, wallClockTimeList
````

````python
class D3QN(D3QN):
    def trainNetwork(self, experiences, epochs=1):
        # this method trains the value network epoch number of times and is called by the trainAgent function
        # it essentially uses the experiences to calculate target, using the targets it calculates the error, which
        # is further used for calulating the loss. It then uses the optimizer over the loss
        # to update the params of the network by backpropagating through the network
        # this function does not return anything
        # you can try out other loss functions other than MSE like Huber loss, MAE, etc.

        #Your code goes in here
        states, actions, rewards, nextStates, dones = self.buffer.splitExperiences(experiences)
        with torch.no_grad():
            x = self.onlineNet(nextStates).argmax(dim=1)
            Q_next = self.targetNet(nextStates)
            mQ_next = Q_next.gather(1, x.unsqueeze(1)).squeeze(1)
            targets = rewards + self.gamma * mQ_next * (1 - dones)
        Q_vals = self.onlineNet(states)
        Q_vals = Q_vals.gather(1, actions.unsqueeze(1)).squeeze(1)
        loss = self.loss(Q_vals, targets)
        self.optimizer.zero_grad()
        loss.backward()
        self.optimizer.step()
        return
````

````python
class D3QN(D3QN):
    def updateNetwork(self, onlineNet, targetNet):
        tau = self.tau
        for o, t in zip(onlineNet.parameters(), targetNet.parameters()):
            t.data.copy_(tau*o.data + (1-tau)*t.data)
        return
````

````python
class D3QN(D3QN):
    def evaluateAgent(self):
        for i in range(self.maxEvalEpisodes):
            state, info = self.env.reset(seed=self.seed + i)
            self.currentEvalReward = 0
            done = False
            while not done:
                action = self.strategyEval(self.onlineNet, state)
                state, r, term, trunc, _ = self.env.step(action)
                done = term or trunc
                self.currentEvalReward += r
            self.performBookKeeping(train=False)
            
        finalEvalRewardsList = self.evalRewards
        return finalEvalRewardsList
````

````python
class D3QN_PER():
    def __init__(self,env, seed, gamma, tau, alpha, beta, beta_rate,
                 bufferSize,
                 batchSize,
                 optimizerFn,
                 optimizerLR,
                 MAX_TRAIN_EPISODES, MAX_EVAL_EPISODES,
                 explorationStrategyTrainFn,
                 explorationStrategyEvalFn,
                 updateFrequency, loss=F.mse_loss):
        self.env = env
        self.seed = seed
        self.gamma = gamma
        self.batchSize = batchSize
        self.maxTrainEpisodes = MAX_TRAIN_EPISODES
        self.maxEvalEpisodes = MAX_EVAL_EPISODES
        self.strategyTrain = explorationStrategyTrainFn
        self.strategyEval = explorationStrategyEvalFn
        self.updateFrequency = updateFrequency
        self.loss = loss
        inDim = env.observation_space.shape[0]
        outDim = env.action_space.n
        self.onlineNet = createDuelingNetwork(inDim, outDim)
        self.targetNet = createDuelingNetwork(inDim, outDim)
        self.targetNet.load_state_dict(self.onlineNet.state_dict())
        self.targetNet.eval()
        self.optimizer = optimizerFn(self.onlineNet.parameters(), lr=optimizerLR)
        self.buffer = ReplayBuffer(bufferSize, bufferType='PER-D3QN', alpha=alpha,beta=beta,beta_rate=beta_rate)
        self.tau = tau
        self.alpha = alpha
        self.beta = beta
        self.beta_rate = beta_rate
        self.initBookKeeping()
````

````python
class D3QN_PER(D3QN_PER):
    def initBookKeeping(self):
        self.trainRewards = []
        self.evalRewards = []
        self.trainTime = []
        self.wallTime = []
        self.totalSteps = []
        self.stepCount = 0
        self.currentEpisodeReward = 0
        self.currentEpisodeSteps = 0
        self.currentTrainTime = 0
        self.currentWallTime = 0
        self.currentEvalReward = 0
        
        return
````

````python
class D3QN_PER(D3QN_PER):
    def trainAgent(self):
        self.updateNetwork(self.onlineNet, self.targetNet)
        evalRewardsList = []
        self.currentTrainTime = 0
        self.currentWallTime = 0
        for e in range(self.maxTrainEpisodes):
            wallStart = time.time()
            state, info = self.env.reset(seed=self.seed+e)
            done = False
            self.currentEpisodeReward = 0
            self.currentEpisodeSteps = 0
            while not done: 
                action = self.strategyTrain(self.onlineNet,state)
                nextState, reward, term, trunc, _ = self.env.step(action)
                done = term or trunc
                self.buffer.store((state,action,reward,nextState,done))
                state = nextState
                self.currentEpisodeReward += reward
                self.currentEpisodeSteps +=1 
                self.stepCount+=1
                if self.buffer.length() >= self.batchSize:
                    experiences, inds, ws = self.buffer.sample(self.batchSize)
                    trainStart = time.time()
                    self.trainNetwork(experiences, inds, ws)
                    trainEnd = time.time()
                    self.currentTrainTime += trainEnd - trainStart         
                if self.stepCount % self.updateFrequency == 0:
                    self.updateNetwork(self.onlineNet, self.targetNet)

            ls = np.array(self.evaluateAgent()[-self.maxEvalEpisodes:])
            evalRewardsList.append(ls.mean())
            wallEnd = time.time()
            self.currentWallTime += wallEnd-wallStart
            self.performBookKeeping(train=True)
            
        trainRewardsList=self.trainRewards
        trainTimeList = self.trainTime
        wallClockTimeList = self.wallTime
        
        return trainRewardsList, trainTimeList, evalRewardsList, wallClockTimeList
````

````python
class D3QN_PER(D3QN_PER):
    def trainNetwork(self, experiences, inds, ws):
        states, actions, rewards, nextStates, dones = self.buffer.splitExperiences(experiences)
        with torch.no_grad():
            x = self.onlineNet(nextStates).argmax(dim=1)
            Q_next = self.targetNet(nextStates)
            mQ_next = Q_next.gather(1, x.unsqueeze(1)).squeeze(1)
            targets = rewards + self.gamma * mQ_next * (1 - dones)
        Q_vals = self.onlineNet(states)
        Q_vals = Q_vals.gather(1, actions.unsqueeze(1)).squeeze(1)
        errors = torch.abs(targets - Q_vals).detach().numpy()
        self.buffer.update(inds, errors)
        loss = (ws * self.loss(Q_vals, targets, reduction='none')).mean()
        self.optimizer.zero_grad()
        loss.backward()
        self.optimizer.step()
        return
````

````python
class D3QN_PER(D3QN_PER):
    def updateNetwork(self, onlineNet, targetNet):
        tau = self.tau
        for o, t in zip(onlineNet.parameters(), targetNet.parameters()):
            t.data.copy_(tau*o.data + (1-tau)*t.data)
        return
````

````python
class D3QN_PER(D3QN_PER):
    def evaluateAgent(self):
        for i in range(self.maxEvalEpisodes):
            state, info = self.env.reset(seed=self.seed + i)
            self.currentEvalReward = 0
            done = False
            while not done:
                action = self.strategyEval(self.onlineNet, state)
                state, r, term, trunc, _ = self.env.step(action)
                done = term or trunc
                self.currentEvalReward += r
            self.performBookKeeping(train=False)
            
        finalEvalRewardsList = self.evalRewards
        return finalEvalRewardsList
````
