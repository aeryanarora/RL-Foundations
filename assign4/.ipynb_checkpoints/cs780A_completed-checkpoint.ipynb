# Helper Functions
<a id="helper"></a>

Write all the helper functions that will be used for value-based and policy based algorithms below. In case you want to add more helper functions, please feel free to add.

````python
def selectGreedyAction(net, state):
    with torch.no_grad():
        stateTensor = torch.as_tensor(state, dtype=torch.float32, device=DEVICE)
        if stateTensor.dim() == 1:
            stateTensor = stateTensor.unsqueeze(0)
        Q = net(stateTensor).squeeze(0).cpu()
    ls = [a for a in range(Q.shape[0]) if Q[a] == torch.max(Q)]
    greedyAction = int(np.random.choice(ls))
    return greedyAction
````

````python
def selectEpsilonGreedyAction(net, state, epsilon):
    # this function gets q-values via the network and selects an action from q-values using epsilon greedy strategy
    # and returns it
    # note this function can be used for decaying epsilon greedy strategy,
    # you would need to create a wrapper function that will handle decaying epsilon
    # you can create this wrapper in this helper function section
    # for the agents you would be implementing it would be nice to play with decaying parameter to get optimal results
    with torch.no_grad():
        stateTensor = torch.as_tensor(state, dtype=torch.float32, device=DEVICE)
        if stateTensor.dim() == 1:
            stateTensor = stateTensor.unsqueeze(0)
        Q = net(stateTensor).squeeze(0).cpu()

    if np.random.rand() < epsilon:
        eGreedyAction = int(np.random.choice(range(Q.shape[0])))
    else:
        ls = [a for a in range(Q.shape[0]) if Q[a] == torch.max(Q)]
        eGreedyAction = int(np.random.choice(ls))

    return eGreedyAction
````

````python
def selectSoftMaxAction(net, state, temp):
    # this function gets q-values via the network and selects an action from q-values using softmax strategy
    # and returns it
    # note this function can be used for decaying temperature softmax strategy,
    # you would need to create a wrapper function that will handle decaying temperature
    # you can create this wrapper in this helper function section
    # for the agents you would be implementing it would be nice to play with decaying parameter to get optimal results
    with torch.no_grad():
        stateTensor = torch.as_tensor(state, dtype=torch.float32, device=DEVICE)
        if stateTensor.dim() == 1:
            stateTensor = stateTensor.unsqueeze(0)
        Q = net(stateTensor).squeeze(0).cpu()
    temp = max(float(temp), 1e-6)
    probs = F.softmax(Q / temp, dim=0).numpy()
    softAction = int(np.random.choice(range(Q.shape[0]), p=probs))
    return softAction
````

````python
# Value Network
def createValueNetwork(inDim, outDim, hDim=[32, 32], activation=F.relu):
    # this creates a Feed Forward Neural Network class and instantiates it and returns the class
    # the class should be derived from torch nn.Module and it should have init and forward method at the very least
    # the forward function should return q-value for each possible action
    class ValueNetwork(nn.Module):
        def __init__(self, inDim, outDim, hDim=[32, 32], activation=F.relu):
            super().__init__()
            dims = [inDim] + list(hDim) + [outDim]
            self.layers = nn.ModuleList(
                [nn.Linear(dims[i], dims[i + 1]) for i in range(len(dims) - 1)]
            )
            self.activation = activation

        def forward(self, state):
            squeeze_back = False
            if state.dim() == 1:
                state = state.unsqueeze(0)
                squeeze_back = True
            x = state
            for i in range(len(self.layers) - 1):
                x = self.activation(self.layers[i](x))
            x = self.layers[-1](x)
            return x.squeeze(0) if squeeze_back else x

    valueNetwork = ValueNetwork(inDim, outDim, hDim, activation).to(DEVICE)
    return valueNetwork
````

````python
# Dueling Network
def createDuelingNetwork(inDim, outDim, hDim=[32, 32], activation=F.relu):
    class DuelNetwork(nn.Module):
        def __init__(self, inDim, outDim, hDim, activation):
            super().__init__()
            dims = [inDim] + list(hDim)
            self.shared = nn.ModuleList(
                [nn.Linear(dims[i], dims[i + 1]) for i in range(len(dims) - 1)]
            )
            self.v_layer = nn.Linear(dims[-1], 1)
            self.a_layer = nn.Linear(dims[-1], outDim)
            self.activation = activation

        def forward(self, state):
            squeeze_back = False
            if state.dim() == 1:
                state = state.unsqueeze(0)
                squeeze_back = True
            x = state
            for layer in self.shared:
                x = self.activation(layer(x))
            v = self.v_layer(x)
            a = self.a_layer(x)
            q = v + (a - a.mean(dim=1, keepdim=True))
            return q.squeeze(0) if squeeze_back else q

    duelNetwork = DuelNetwork(inDim, outDim, hDim, activation).to(DEVICE)
    return duelNetwork
````

````python
# Policy Network
def createPolicyNetwork(inDim, outDim, hDim=[32, 32], activation=F.relu):
    class PolicyNetwork(nn.Module):
        def __init__(self, inDim, outDim, hDim=[32, 32], activation=F.relu):
            super().__init__()
            dims = [inDim] + list(hDim) + [outDim]
            self.layers = nn.ModuleList(
                [nn.Linear(dims[i], dims[i + 1]) for i in range(len(dims) - 1)]
            )
            self.activation = activation

        def forward(self, state):
            squeeze_back = False
            if state.dim() == 1:
                state = state.unsqueeze(0)
                squeeze_back = True
            x = state
            for i in range(len(self.layers) - 1):
                x = self.activation(self.layers[i](x))
            x = self.layers[-1](x)
            return x.squeeze(0) if squeeze_back else x

    policyNetwork = PolicyNetwork(inDim, outDim, hDim, activation).to(DEVICE)
    return policyNetwork
````

````python
def huberLoss(error, delta):
    # this function calculates the huber loss for the error using the delta parameter
    abs_error = torch.abs(error)
    loss = torch.where(abs_error <= delta, 0.5 * error ** 2, delta * (abs_error - 0.5 * delta))
    return loss
````

````python
# in case you want to create any other helper function, the code goes in here

def set_global_seed(seed):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)

def make_env(env_name, seed=None):
    env = gym.make(env_name)
    if seed is not None:
        env.reset(seed=seed)
        try:
            env.action_space.seed(seed)
        except Exception:
            pass
    return env

def linear_decay(start_value, end_value, current_step, decay_steps):
    if decay_steps <= 0:
        return end_value
    fraction = min(max(current_step / decay_steps, 0.0), 1.0)
    return start_value + fraction * (end_value - start_value)

def make_epsilon_greedy_wrapper(start_epsilon=1.0, end_epsilon=0.05, decay_episodes=100):
    def wrapped(net, state, episode_idx=0):
        eps = linear_decay(start_epsilon, end_epsilon, episode_idx, decay_episodes)
        return selectEpsilonGreedyAction(net, state, eps)
    return wrapped

def make_softmax_wrapper(start_temp=1.0, end_temp=0.1, decay_episodes=100):
    def wrapped(net, state, episode_idx=0):
        current_temp = linear_decay(start_temp, end_temp, episode_idx, decay_episodes)
        return selectSoftMaxAction(net, state, current_temp)
    return wrapped

def gather_q_values(q_values, actions):
    return q_values.gather(1, actions.unsqueeze(1)).squeeze(1)

def unpack_step(step_output):
    next_state, reward, terminated, truncated, info = step_output
    done = terminated or truncated
    return next_state, reward, done, info
````

````python
# in case you want to create any other helper function, the code goes in here

def plotAcrossAgents(quantityByAgent, totalEpisodeCount, ylabel, title):
    plt.figure(figsize=(9, 5))
    episodes = np.arange(1, totalEpisodeCount + 1)
    for agentName, quantityListDict in quantityByAgent.items():
        data = np.array(list(quantityListDict.values()), dtype=np.float32)
        mean = data.mean(axis=0)
        low = data.min(axis=0)
        high = data.max(axis=0)
        plt.plot(episodes, mean, linewidth=2, label=agentName)
        plt.fill_between(episodes, low, high, alpha=0.12)
    plt.xlabel("Episodes")
    plt.ylabel(ylabel)
    plt.title(title)
    plt.legend(bbox_to_anchor=(1.02, 1), loc="upper left")
    plt.tight_layout()
    plt.show()

def evaluate_greedy_policy(env, net, max_eval_episodes):
    rewards = []
    for eval_idx in range(max_eval_episodes):
        state, _ = env.reset()
        done = False
        total_reward = 0.0
        while not done:
            action = selectGreedyAction(net, state)
            state, reward, done, _ = unpack_step(env.step(action))
            total_reward += reward
        rewards.append(total_reward)
    return rewards
````

## ReplayBuffer (50 Marks)

In next few cells, you will implement replaybuffer class.

This class creates a buffer for storing and retrieving experiences. This is a generic class and can be used
for different agents like NFQ, DQN, DDQN, PER_DDQN, etc.
Following are the methods for this class which are implemented in subsequent cells

```
class ReplayBuffer():
    def __init__(self, bufferSize, **kwargs)
    def store(self, experience)
    def update(self, indices, priorities)
    def collectExperiences(env, state, explorationStrategy, net = None)
    def sample(self, batchSize, **kwargs)
    def splitExperiences(self, experiences)
    def length(self)
```   

````python
class ReplayBuffer():
    def __init__(self, bufferSize, bufferType='DQN', **kwargs):
        # this function creates the relevant data-structures, and intializes all relevant variables
        # it can take variable number of parameters like alpha, beta, beta_rate (required for PER)
        # here the bufferType variable can be used to maintain one class for all types of agents
        # using the bufferType parameter in the methods below, you can implement all possible functionalities
        # that could be used for different types of agents
        # permissible values for bufferType = NFQ, DQN, DDQN, D3QN and PER-D3QN
        self.bufferSize = int(bufferSize)
        self.bufferType = bufferType
        self.buffer = deque(maxlen=self.bufferSize)
        if bufferType == 'PER-D3QN':
            self.alpha = kwargs.get('alpha', alpha)
            self.beta = kwargs.get('beta', beta)
            self.beta_rate = kwargs.get('beta_rate', beta_rate)
            self.priorities = deque(maxlen=self.bufferSize)
            self.eps = 1e-6
        return
````

````python
class ReplayBuffer(ReplayBuffer):
    def store(self, experience):
        # stores the experiences, based on parameters in init it can assign priorities, etc.
        # this function does not return anything
        self.buffer.append(experience)
        if self.bufferType == 'PER-D3QN':
            max_priority = max(self.priorities) if len(self.priorities) > 0 else 1.0
            self.priorities.append(float(max_priority))
        return
````

````python
class ReplayBuffer(ReplayBuffer):
    def update(self, indices, priorities):
        if self.bufferType != 'PER-D3QN':
            return
        priorities = np.asarray(priorities, dtype=np.float32)
        for i, idx in enumerate(indices):
            self.priorities[idx] = float(max(priorities[i], self.eps))
        return
````

````python
class ReplayBuffer(ReplayBuffer):
    def collectExperiences(self, env, state, explorationStrategy, countExperiences, net=None):
        for _ in range(countExperiences):
            action = explorationStrategy(net, state)
            next_state, reward, done, _ = unpack_step(env.step(action))
            experience = (state, action, reward, next_state, done)
            self.store(experience)
            state = next_state
            if done:
                break
        return
````

````python
class ReplayBuffer(ReplayBuffer):
    def sample(self, batchSize, **kwargs):
        # this method returns batchSize number of experiences
        # based on extra arguments, it could do sampling or it could return the latest batchSize experiences or
        # via some other strategy
        # in the case of Prioritized Experience Replay (PER) the sampling needs to take into account the priorities
        # this function returns experience samples
        batchSize = min(int(batchSize), len(self.buffer))
        if batchSize == 0:
            return []

        if self.bufferType == 'PER-D3QN':
            priorities = np.array(self.priorities, dtype=np.float32)
            scaled = priorities ** self.alpha
            probs = scaled / scaled.sum()
            indices = np.random.choice(len(self.buffer), size=batchSize, replace=False, p=probs)
            experiencesList = [self.buffer[idx] for idx in indices]
            weights = (len(self.buffer) * probs[indices]) ** (-self.beta)
            weights = weights / weights.max()
            self.beta = min(1.0, self.beta * self.beta_rate)
            return experiencesList, indices, torch.as_tensor(weights, dtype=torch.float32, device=DEVICE)

        experiencesList = random.sample(list(self.buffer), k=batchSize)
        return experiencesList
````

````python
class ReplayBuffer(ReplayBuffer):
    def splitExperiences(self, experiences):
        # it takes in experiences and gives the following:
        # states, actions, rewards, nextStates, dones
        states = torch.as_tensor(np.array([e[0] for e in experiences]), dtype=torch.float32, device=DEVICE)
        actions = torch.as_tensor(np.array([e[1] for e in experiences]), dtype=torch.long, device=DEVICE)
        rewards = torch.as_tensor(np.array([e[2] for e in experiences]), dtype=torch.float32, device=DEVICE)
        nextStates = torch.as_tensor(np.array([e[3] for e in experiences]), dtype=torch.float32, device=DEVICE)
        dones = torch.as_tensor(np.array([e[4] for e in experiences]), dtype=torch.float32, device=DEVICE)
        return states, actions, rewards, nextStates, dones
````

````python
class ReplayBuffer(ReplayBuffer):
    def length(self):
        bufferSize = len(self.buffer)
        return bufferSize
````

## Neural Fitted Q (NFQ) (50 Marks)

Implement the Neural Fitted Q algorithm. We have studied about NFQ algorithm in the Lecture. Use the function definitions (given below).

class NFQ():

````python
class NFQ():
    def __init__(self, env, seed, gamma, epochs,
                 bufferSize,
                 batchSize,
                 optimizerFn,
                 optimizerLR,
                 MAX_TRAIN_EPISODES, MAX_EVAL_EPISODES,
                 explorationStrategyTrainFn,
                 explorationStrategyEvalFn):
        self.env = env
        self.seed = seed
        set_global_seed(seed)
        self.env.reset(seed=seed)
        self.net = createValueNetwork(env.observation_space.shape[0], env.action_space.n)
        self.gamma = gamma
        self.epochs = epochs
        self.buffer = ReplayBuffer(bufferSize, bufferType='NFQ')
        self.strategyTrain = explorationStrategyTrainFn
        self.strategyEval = explorationStrategyEvalFn
        self.maxTrainEpisodes = MAX_TRAIN_EPISODES
        self.maxEvalEpisodes = MAX_EVAL_EPISODES
        self.optimizer = optimizerFn(self.net.parameters(), lr=optimizerLR)
        self.batchSize = batchSize
        self.initBookKeeping()
````

````python
class NFQ(NFQ):
    def initBookKeeping(self):
        # this method creates and intializes all the variables required for book-keeping values and it is called
        # init method
        self.trainRewards = []
        self.evalRewards = []
        self.trainTime = []
        self.wallTime = []
        self.totalSteps = []
        self.stepCount = 0
        self.currentEpisodeReward = 0.0
        self.currentEpisodeSteps = 0
        self.currentTrainTime = 0.0
        self.currentWallTime = 0.0
        self.currentEvalReward = 0.0
````

````python
class NFQ(NFQ):
    def performBookKeeping(self, train=True):
        # this method updates relevant variables for the bookKeeping
        if train:
            self.trainRewards.append(float(self.currentEpisodeReward))
            self.trainTime.append(float(self.currentTrainTime))
            self.wallTime.append(float(self.currentWallTime))
            self.totalSteps.append(int(self.currentEpisodeSteps))
        else:
            self.evalRewards.append(float(self.currentEvalReward))
````

````python
class NFQ(NFQ):
    def runNFQ(self):
        # this is the main method, it trains the agent, performs bookkeeping while training and finally evaluates
        trainRewardsList, trainTimeList, evalRewardsList, wallClockTimeList = self.trainAgent()
        finalEvalReward = float(np.mean(self.evaluateAgent()))
        return trainRewardsList, trainTimeList, evalRewardsList, wallClockTimeList, finalEvalReward
````

````python
class NFQ(NFQ):
    def trainAgent(self):
        # this method collects experiences and trains the NFQ agent and does BookKeeping while training.
        train_start = time.perf_counter()
        cumulative_train_time = 0.0

        for episode in range(self.maxTrainEpisodes):
            state, _ = self.env.reset(seed=self.seed + episode)
            done = False
            self.currentEpisodeReward = 0.0
            self.currentEpisodeSteps = 0

            episode_train_start = time.perf_counter()
            while not done:
                action = self.strategyTrain(self.net, state, episode)
                next_state, reward, done, _ = unpack_step(self.env.step(action))
                self.buffer.store((state, action, reward, next_state, done))
                state = next_state

                self.currentEpisodeReward += reward
                self.currentEpisodeSteps += 1
                self.stepCount += 1

            experiences = list(self.buffer.buffer)
            self.trainNetwork(experiences, self.epochs)

            cumulative_train_time += time.perf_counter() - episode_train_start
            self.currentTrainTime = cumulative_train_time
            self.currentWallTime = time.perf_counter() - train_start
            self.performBookKeeping(train=True)

            eval_rewards = self.evaluateAgent()
            self.currentEvalReward = float(np.mean(eval_rewards))
            self.performBookKeeping(train=False)

        trainRewardsList = self.trainRewards
        trainTimeList = self.trainTime
        evalRewardsList = self.evalRewards
        wallClockTimeList = self.wallTime
        return trainRewardsList, trainTimeList, evalRewardsList, wallClockTimeList
````

````python
class NFQ(NFQ):
    def trainNetwork(self, experiences, epochs):
        # this method trains the value network epoch number of times and is called by the trainAgent function
        # it essentially uses the experiences to calculate target, using the targets it calculates the error, which
        # is further used for calulating the loss. It then uses the optimizer over the loss
        # to update the params of the network by backpropagating through the network
        if len(experiences) == 0:
            return

        states, actions, rewards, nextStates, dones = self.buffer.splitExperiences(experiences)
        for _ in range(epochs):
            q_vals = self.net(states)
            chosen_q = gather_q_values(q_vals, actions)
            with torch.no_grad():
                next_q = self.net(nextStates).max(dim=1).values
                targets = rewards + self.gamma * next_q * (1.0 - dones)

            error = chosen_q - targets
            loss = huberLoss(error, delta).mean()

            self.optimizer.zero_grad()
            loss.backward()
            self.optimizer.step()
        return
````

````python
class NFQ(NFQ):
    def evaluateAgent(self):
        # this function evaluates the agent using the value network, it evaluates agent for MAX_EVAL_EPISODES
        finalEvalRewardsList = evaluate_greedy_policy(self.env, self.net, self.maxEvalEpisodes)
        return finalEvalRewardsList
````

````python
class D3QN():
    def __init__(self, env, seed, gamma, tau,
                 bufferSize,
                 batchSize,
                 optimizerFn,
                 optimizerLR,
                 MAX_TRAIN_EPISODES, MAX_EVAL_EPISODES,
                 explorationStrategyTrainFn,
                 explorationStrategyEvalFn,
                 updateFrequency):
        # this D3QN method
        # 1. creates and initializes (with seed) the environment, train/eval episodes, gamma, batch size, MAX_TRAIN_EPISODES, MAX_EVAL_EPISODES, updateFrequency etc.
        # 2. creates and intializes all the variables required for book-keeping values via the initBookKeeping method
        # 3. creates target and online Q-networks using the createDuelingNetwork above
        # 4. creates and initializes (with network params) the optimizer function
        # 5. sets the explorationStartegy variables/functions for train and evaluation
        # 6. sets the batchSize for the number of experiences
        # 7. Creates the replayBuffer
        self.env = env
        self.seed = seed
        set_global_seed(seed)
        self.env.reset(seed=seed)
        self.gamma = gamma
        self.tau = tau
        self.onlineNet = createDuelingNetwork(env.observation_space.shape[0], env.action_space.n)
        self.targetNet = createDuelingNetwork(env.observation_space.shape[0], env.action_space.n)
        self.targetNet.load_state_dict(self.onlineNet.state_dict())
        self.targetNet.eval()
        self.buffer = ReplayBuffer(bufferSize, bufferType='D3QN')
        self.batchSize = batchSize
        self.optimizer = optimizerFn(self.onlineNet.parameters(), lr=optimizerLR)
        self.maxTrainEpisodes = MAX_TRAIN_EPISODES
        self.maxEvalEpisodes = MAX_EVAL_EPISODES
        self.strategyTrain = explorationStrategyTrainFn
        self.strategyEval = explorationStrategyEvalFn
        self.updateFrequency = updateFrequency
        self.initBookKeeping()
````

````python
class D3QN(D3QN):
    def initBookKeeping(self):
        # this method creates and intializes all the variables required for book-keeping values and it is called
        self.trainRewards = []
        self.evalRewards = []
        self.trainTime = []
        self.wallTime = []
        self.totalSteps = []
        self.stepCount = 0
        self.currentEpisodeReward = 0.0
        self.currentEpisodeSteps = 0
        self.currentTrainTime = 0.0
        self.currentWallTime = 0.0
        self.currentEvalReward = 0.0
        return
````

````python
class D3QN(D3QN):
    def performBookKeeping(self, train=True):
        # this method updates relevant variables for the bookKeeping
        if train:
            self.trainRewards.append(float(self.currentEpisodeReward))
            self.trainTime.append(float(self.currentTrainTime))
            self.wallTime.append(float(self.currentWallTime))
            self.totalSteps.append(int(self.currentEpisodeSteps))
        else:
            self.evalRewards.append(float(self.currentEvalReward))
        return
````

````python
class D3QN(D3QN):
    def runD3QN(self):
        # this is the main method, it trains the agent, performs bookkeeping while training and finally evaluates
        trainRewardsList, trainTimeList, evalRewardsList, wallClockTimeList = self.trainAgent()
        finalEvalReward = float(np.mean(self.evaluateAgent()))
        return trainRewardsList, trainTimeList, evalRewardsList, wallClockTimeList, finalEvalReward
````

````python
class D3QN(D3QN):
    def trainAgent(self):
        # this method collects experiences and trains the agent and does BookKeeping while training.
        train_start = time.perf_counter()
        cumulative_train_time = 0.0

        for episode in range(self.maxTrainEpisodes):
            state, _ = self.env.reset(seed=self.seed + episode)
            done = False
            self.currentEpisodeReward = 0.0
            self.currentEpisodeSteps = 0
            episode_train_start = time.perf_counter()

            while not done:
                action = self.strategyTrain(self.onlineNet, state, episode)
                next_state, reward, done, _ = unpack_step(self.env.step(action))
                self.buffer.store((state, action, reward, next_state, done))
                state = next_state

                self.currentEpisodeReward += reward
                self.currentEpisodeSteps += 1
                self.stepCount += 1

                if self.buffer.length() >= self.batchSize:
                    experiences = self.buffer.sample(self.batchSize)
                    self.trainNetwork(experiences, GRADIENT_EPOCHS)
                    if self.stepCount % self.updateFrequency == 0:
                        self.updateNetwork(self.onlineNet, self.targetNet)

            cumulative_train_time += time.perf_counter() - episode_train_start
            self.currentTrainTime = cumulative_train_time
            self.currentWallTime = time.perf_counter() - train_start
            self.performBookKeeping(train=True)

            eval_rewards = self.evaluateAgent()
            self.currentEvalReward = float(np.mean(eval_rewards))
            self.performBookKeeping(train=False)

        trainRewardsList = self.trainRewards
        trainTimeList = self.trainTime
        evalRewardsList = self.evalRewards
        wallClockTimeList = self.wallTime
        return trainRewardsList, trainTimeList, evalRewardsList, wallClockTimeList
````

````python
class D3QN(D3QN):
    def trainNetwork(self, experiences, epochs):
        # this method trains the value network epoch number of times and is called by the trainAgent function
        if len(experiences) == 0:
            return

        states, actions, rewards, nextStates, dones = self.buffer.splitExperiences(experiences)
        for _ in range(epochs):
            q_vals = self.onlineNet(states)
            chosen_q = gather_q_values(q_vals, actions)

            with torch.no_grad():
                next_actions = self.onlineNet(nextStates).argmax(dim=1)
                next_q = gather_q_values(self.targetNet(nextStates), next_actions)
                targets = rewards + self.gamma * next_q * (1.0 - dones)

            error = chosen_q - targets
            loss = huberLoss(error, delta).mean()

            self.optimizer.zero_grad()
            loss.backward()
            self.optimizer.step()
        return
````

````python
class D3QN(D3QN):
    def updateNetwork(self, onlineNet, targetNet):
        # this function updates the onlineNetwork with the target network using Polyak averaging
        with torch.no_grad():
            for target_param, online_param in zip(targetNet.parameters(), onlineNet.parameters()):
                target_param.data.copy_(self.tau * online_param.data + (1.0 - self.tau) * target_param.data)
        return
````

````python
class D3QN(D3QN):
    def evaluateAgent(self):
        # this function evaluates the agent using the value network, it evaluates agent for MAX_EVAL_EPISODES
        finalEvalRewardsList = evaluate_greedy_policy(self.env, self.onlineNet, self.maxEvalEpisodes)
        return finalEvalRewardsList
````

````python
class D3QN_PER(D3QN_PER):
    def trainAgent(self):
        # this method collects experiences and trains the agent and does BookKeeping while training.
        train_start = time.perf_counter()
        cumulative_train_time = 0.0

        for episode in range(self.maxTrainEpisodes):
            state, _ = self.env.reset(seed=self.seed + episode)
            done = False
            self.currentEpisodeReward = 0.0
            self.currentEpisodeSteps = 0
            episode_train_start = time.perf_counter()

            while not done:
                action = self.strategyTrain(self.onlineNet, state, episode)
                next_state, reward, done, _ = unpack_step(self.env.step(action))
                self.buffer.store((state, action, reward, next_state, done))
                state = next_state

                self.currentEpisodeReward += reward
                self.currentEpisodeSteps += 1
                self.stepCount += 1

                if self.buffer.length() >= self.batchSize:
                    experiences = self.buffer.sample(self.batchSize)
                    self.trainNetwork(experiences)
                    if self.stepCount % self.updateFrequency == 0:
                        self.updateNetwork(self.onlineNet, self.targetNet)

            cumulative_train_time += time.perf_counter() - episode_train_start
            self.currentTrainTime = cumulative_train_time
            self.currentWallTime = time.perf_counter() - train_start
            self.performBookKeeping(train=True)

            eval_rewards = self.evaluateAgent()
            self.currentEvalReward = float(np.mean(eval_rewards))
            self.performBookKeeping(train=False)

        trainRewardsList = self.trainRewards
        trainTimeList = self.trainTime
        evalRewardsList = self.evalRewards
        wallClockTimeList = self.wallTime
        return trainRewardsList, trainTimeList, evalRewardsList, wallClockTimeList
````

````python
class D3QN_PER(D3QN_PER):
    def trainNetwork(self, experiences):
        # this method trains the value network epoch number of times and is called by the trainAgent function
        if len(experiences) == 0:
            return

        sampled_experiences, indices, weights = experiences
        states, actions, rewards, nextStates, dones = self.buffer.splitExperiences(sampled_experiences)

        q_vals = self.onlineNet(states)
        chosen_q = gather_q_values(q_vals, actions)

        with torch.no_grad():
            next_actions = self.onlineNet(nextStates).argmax(dim=1)
            next_q = gather_q_values(self.targetNet(nextStates), next_actions)
            targets = rewards + self.gamma * next_q * (1.0 - dones)

        td_error = chosen_q - targets
        elementwise_loss = huberLoss(td_error, delta)
        loss = (weights * elementwise_loss).mean()

        self.optimizer.zero_grad()
        loss.backward()
        self.optimizer.step()

        new_priorities = torch.abs(td_error).detach().cpu().numpy() + 1e-6
        self.buffer.update(indices, new_priorities)
        return
````

````python
class D3QN_PER(D3QN_PER):
    def updateNetwork(self, onlineNet, targetNet):
        # this function updates the onlineNetwork with the target network using Polyak averaging
        with torch.no_grad():
            for target_param, online_param in zip(targetNet.parameters(), onlineNet.parameters()):
                target_param.data.copy_(self.tau * online_param.data + (1.0 - self.tau) * target_param.data)
        return
````

````python
class D3QN_PER(D3QN_PER):
    def evaluateAgent(self):
        # this function evaluates the agent using the value network, it evaluates agent for MAX_EVAL_EPISODES
        finalEvalRewardsList = evaluate_greedy_policy(self.env, self.onlineNet, self.maxEvalEpisodes)
        return finalEvalRewardsList
````

````python
def runDeepValueBasedAgents():
    # this function will initialize 5 different instances of the env (using different seeds), run all the agents
    # over these different instances. Collects results and generate the plots stated above.

    all_results = {}
    env_names = ["CartPole-v0", "MountainCar-v0"]

    for env_name in env_names:
        cfg = ENV_CONFIGS[env_name]
        total_episodes = cfg["train_episodes"]

        agent_builders = {
            "NFQ": lambda env, seed: NFQ(
                env, seed, gamma, cfg["nfq_epochs"],
                cfg["buffer_size"], cfg["batch_size"],
                optim.Adam, cfg["optimizer_lr"],
                cfg["train_episodes"], cfg["eval_episodes"],
                make_epsilon_greedy_wrapper(cfg["epsilon_start"], cfg["epsilon_end"], cfg["epsilon_decay_episodes"]),
                selectGreedyAction
            ),
            "DQN": lambda env, seed: DQN(
                env, seed, gamma,
                cfg["buffer_size"], cfg["batch_size"],
                optim.Adam, cfg["optimizer_lr"],
                cfg["train_episodes"], cfg["eval_episodes"],
                make_epsilon_greedy_wrapper(cfg["epsilon_start"], cfg["epsilon_end"], cfg["epsilon_decay_episodes"]),
                selectGreedyAction,
                cfg["update_frequency"]
            ),
            "DDQN": lambda env, seed: DDQN(
                env, seed, gamma,
                cfg["buffer_size"], cfg["batch_size"],
                optim.Adam, cfg["optimizer_lr"],
                cfg["train_episodes"], cfg["eval_episodes"],
                make_epsilon_greedy_wrapper(cfg["epsilon_start"], cfg["epsilon_end"], cfg["epsilon_decay_episodes"]),
                selectGreedyAction,
                cfg["update_frequency"]
            ),
            "D3QN": lambda env, seed: D3QN(
                env, seed, gamma, cfg["tau"],
                cfg["buffer_size"], cfg["batch_size"],
                optim.Adam, cfg["optimizer_lr"],
                cfg["train_episodes"], cfg["eval_episodes"],
                make_epsilon_greedy_wrapper(cfg["epsilon_start"], cfg["epsilon_end"], cfg["epsilon_decay_episodes"]),
                selectGreedyAction,
                cfg["update_frequency"]
            ),
            "D3QN-PER": lambda env, seed: D3QN_PER(
                env, seed, gamma, cfg["tau"], cfg["alpha"], cfg["beta"], cfg["beta_rate"],
                cfg["buffer_size"], cfg["batch_size"],
                optim.Adam, cfg["optimizer_lr"],
                cfg["train_episodes"], cfg["eval_episodes"],
                make_epsilon_greedy_wrapper(cfg["epsilon_start"], cfg["epsilon_end"], cfg["epsilon_decay_episodes"]),
                selectGreedyAction,
                cfg["update_frequency"]
            ),
        }

        env_results = {}
        for agent_name, builder in agent_builders.items():
            print(f"Running {agent_name} on {env_name}")
            env_results[agent_name] = {
                "train_rewards": {},
                "eval_rewards": {},
                "train_time": {},
                "wall_time": {},
                "steps": {},
                "final_eval": {}
            }

            for seed in SEEDS:
                env = make_env(env_name, seed)
                agent = builder(env, seed)

                if agent_name == "NFQ":
                    train_rewards, train_time, eval_rewards, wall_time, final_eval = agent.runNFQ()
                elif agent_name == "DQN":
                    train_rewards, train_time, eval_rewards, wall_time, final_eval = agent.runDQN()
                elif agent_name == "DDQN":
                    train_rewards, train_time, eval_rewards, wall_time, final_eval = agent.runDDQN()
                elif agent_name == "D3QN":
                    train_rewards, train_time, eval_rewards, wall_time, final_eval = agent.runD3QN()
                else:
                    train_rewards, train_time, eval_rewards, wall_time, final_eval = agent.runD3QN_PER()

                env_results[agent_name]["train_rewards"][seed] = train_rewards
                env_results[agent_name]["eval_rewards"][seed] = eval_rewards
                env_results[agent_name]["train_time"][seed] = train_time
                env_results[agent_name]["wall_time"][seed] = wall_time
                env_results[agent_name]["steps"][seed] = agent.totalSteps
                env_results[agent_name]["final_eval"][seed] = final_eval
                env.close()

        all_results[env_name] = env_results

        metric_specs = [
            ("train_rewards", "Mean Train Reward", f"{env_name}: Mean Train Reward vs Episodes"),
            ("eval_rewards", "Mean Eval Reward", f"{env_name}: Mean Eval Reward vs Episodes"),
            ("steps", "Steps per Episode", f"{env_name}: Steps per Episode"),
            ("train_time", "Train Time (s)", f"{env_name}: Cumulative Train Time vs Episodes"),
            ("wall_time", "Wall Clock Time (s)", f"{env_name}: Wall Clock Time vs Episodes"),
        ]

        for metric_key, ylabel, title in metric_specs:
            quantityByAgent = {agent_name: env_results[agent_name][metric_key] for agent_name in env_results}
            plotAcrossAgents(quantityByAgent, total_episodes, ylabel, title)

        print(f"\nFinal evaluation reward summary for {env_name}")
        for agent_name in env_results:
            final_vals = list(env_results[agent_name]["final_eval"].values())
            print(f"{agent_name:8s}: mean={np.mean(final_vals):8.3f}, min={np.min(final_vals):8.3f}, max={np.max(final_vals):8.3f}")

    return all_results
````
